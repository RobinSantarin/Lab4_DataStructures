```python
from collections import defaultdict

LAST_NAME = "SANTARIN"
FAVORITE_ARTIST = "UNCLE_DAGS"
SEED_NUM = 65

artist_values = [ord(c) for c in FAVORITE_ARTIST if c.isalpha()]

def generated_value(index, multiplier=1):
    return (SEED_NUM + artist_values[index % len(artist_values)]
            * multiplier) % 100 + 1

print("=== STUDENT PERFORMANCE DATA ANALYZER ===")

students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS",
            "PHYSICS", "ENGLISH"]

records = []

for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        score = min(generated_value(i + j, j + 1), 100)
        records.append({
            "student": student,
            "subject": subject,
            "score": score
        })

records.append(records[0].copy())
records.append({
    "student": students[0],
    "subject": "ENGLISH",
    "score": None
})

organized = defaultdict(list)

for r in records:
    organized[r["student"]].append(r)

def performance(score):
    if score is None:
        return "INCOMPLETE"
    if score >= 90:
        return "EXCELLENT"
    if score >= 80:
        return "GOOD"
    if score >= 75:
        return "PASSING"
    return "NEEDS IMPROVEMENT"

print("\nGENERATED STUDENT RECORDS:")
for r in records:
    print(r)

print("\nREPEATED RECORDS:")
seen = set()
for r in records:
    key = (r["student"], r["subject"], r["score"])
    if key in seen:
        print(r)
    seen.add(key)

print("\nINCOMPLETE RECORDS:")
for r in records:
    if r["score"] is None:
        print(r)

print("\nPERFORMANCE CATEGORIES:")
for r in records:
    print(r["student"], "|", r["subject"],
          "| Score:", r["score"],
          "|", performance(r["score"]))

print("\nAVERAGE SCORE BY STUDENT:")
for student, data in organized.items():
    scores = [r["score"] for r in data
              if r["score"] is not None]
    print(student, round(sum(scores) / len(scores), 2))

print("\nAVERAGE SCORE BY SUBJECT:")
for subject in subjects:
    scores = [r["score"] for r in records
              if r["subject"] == subject
              and r["score"] is not None]
    print(subject, round(sum(scores) / len(scores), 2))

print("\nFINAL SUMMARY")
print("Total records:", len(records))
print("Students:", len(students))
print("Subjects:", len(subjects))
print("Incomplete records:",
      sum(r["score"] is None for r in records))
```

```python
from collections import defaultdict, Counter

LAST_NAME = "SANTARIN"
FAVORITE_ARTIST = "UNCLE_DAGS"
SEED_NUM = 65

artist_values = [ord(c) for c in FAVORITE_ARTIST if c.isalpha()]

print("=== TRANSACTION HISTORY ANALYZER ===")

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES",
            "CALCULATOR"]

prices = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 250,
    "HEADPHONES": 350,
    "CALCULATOR": 180
}

transactions = []

for i, customer in enumerate(customers):
    for j, product in enumerate(products):
        quantity = (
            SEED_NUM + artist_values[(i + j) % len(artist_values)]
        ) % 5 + 1

        transactions.append({
            "customer": customer,
            "product": product,
            "quantity": quantity,
            "price": prices[product],
            "total": quantity * prices[product]
        })

transactions.append(transactions[0].copy())

customer_data = defaultdict(list)
product_data = defaultdict(list)
purchase_counts = Counter()

for t in transactions:
    customer_data[t["customer"]].append(t)
    product_data[t["product"]].append(t)
    purchase_counts[(t["customer"], t["product"])] += 1

print("\nGENERATED TRANSACTIONS:")
for t in transactions:
    print(t)

print("\nREPEATED PURCHASES:")
for purchase, count in purchase_counts.items():
    if count > 1:
        print(purchase, "| Count:", count)

print("\nCUSTOMER SPENDING:")
for customer, data in customer_data.items():
    total = sum(r["total"] for r in data)
    print(customer, "| Total spending: ₱", total)

print("\nPRODUCT SUMMARIES:")
for product, data in product_data.items():
    quantity = sum(r["quantity"] for r in data)
    revenue = sum(r["total"] for r in data)
    print(product, "| Quantity:", quantity,
          "| Revenue: ₱", revenue)

print("\nORDERED TRANSACTIONS BY TOTAL:")
ordered = sorted(transactions,
                 key=lambda r: r["total"], reverse=True)
for t in ordered:
    print(t)

print("\nOVERALL TRANSACTION SUMMARY")
print("Total transaction records:", len(transactions))
print("Distinct customer-product pairs:", len(purchase_counts))
print("Total revenue: ₱",
      sum(t["total"] for t in transactions))

most_purchased = max(
    product_data,
    key=lambda p: sum(r["quantity"] for r in product_data[p])
)
print("Most purchased product:", most_purchased)
```
